```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Online_Retail.csv",encoding="latin-1")
```

```python
df.shape
```

```python
df.head()
```

```python
#Handling duplicates
df.duplicated().sum()
df.drop_duplicates(inplace=True,keep="first")
```

```python
df.duplicated().sum()
```

```python
df.isna().sum()
```

```python
#Handling missing values
df['Description']=df['Description'].fillna("unknown")
df['CustomerID']=df['CustomerID'].fillna(0000)
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
#Processing data categories
df['InvoiceDate']=pd.to_datetime(df['InvoiceDate'])
df.dtypes
```

```python
#Split the date column into day, month, and year.
df['year']=df['InvoiceDate'].dt.year
df['month']=df['InvoiceDate'].dt.month
df['day']=df['InvoiceDate'].dt.day
```

```python
df.describe()
```

```python
#Split the quantity column into returned quantity and sold quantity.
df['returns']=df[df['Quantity']<0]['Quantity']*-1
df['returns']=df['returns'].fillna(0)
df['validQuantity']=df[df['Quantity']>0]['Quantity']
df['validQuantity']=df['validQuantity'].fillna(0)
```

```python
df.head()
```

```python
#Determining the number of negative values ​​in the unit price column.
c=df['UnitPrice']<0
c.sum()
```

```python
#Handling negative values ​​in the unit price column
df=df[df['UnitPrice']>0]
```

```python
df.describe()
```

```python
#Separate the unit price column into one for returns and another for actual sales.
df['returns_UnitPrice']=df[df['returns']>0]['UnitPrice']
df['returns_UnitPrice']=df['returns_UnitPrice'].fillna(0)
df['valid_UnitPrice']=df[df['validQuantity']>0]['UnitPrice']
df['valid_UnitPrice']=df['valid_UnitPrice'].fillna(0)
```

```python
#Create a Net Sales Total column.
df['Clean_Revenue']=df['valid_UnitPrice']*df['validQuantity']
```

The most profitable countries:

```python
como=df.groupby("Country")['Clean_Revenue'].sum().sort_values(ascending=False).head(5)
plt.bar(como.index,como.values)
plt.title("Total sales for each country")
plt.xlabel("country")
plt.ylabel("sales")
plt.tight_layout()
plt.xticks(rotation=90)
plt.show()
como
```

Most profitable products:

```python
comn=df.groupby("Description")['Clean_Revenue'].sum().sort_values(ascending=False).head(5)
plt.bar(comn.index,comn.values)
plt.title("Total sales for each Description")
plt.xlabel("Description")
plt.ylabel("sales")
plt.tight_layout()
plt.xticks(rotation=90)
plt.show()
comn
```

Countries with the highest product return rates:

```python
coma=df.groupby("Country")['returns'].sum().sort_values(ascending=False).head(5)
plt.bar(coma.index,coma.values)
plt.title("Total number of returns per country")
plt.xlabel("Country")
plt.ylabel("Number of returns")
plt.tight_layout()
plt.xticks(rotation=90)
plt.show()
coma
```

Most returned products:

```python
cona=df.groupby("Description")['returns'].sum().sort_values(ascending=False).head(5)
plt.bar(cona.index,cona.values)
plt.title("Total number of returns per country")
plt.xlabel("Description")
plt.ylabel("Number of returns")
plt.tight_layout()
plt.xticks(rotation=90)
plt.show()
cona
```

Best-selling products by volume:

```python
oona=df.groupby("Description")['validQuantity'].sum().sort_values(ascending=False).head(5)
plt.bar(oona.index,oona.values)
plt.title("Total number of sales per product")
plt.xlabel("Description")
plt.ylabel("Number of sales")
plt.tight_layout()
plt.xticks(rotation=90)
plt.show()
oona
```
